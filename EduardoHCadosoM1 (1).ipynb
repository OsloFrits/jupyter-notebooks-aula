### Questão 1: Orçamento de viagem (1,0 ponto)

Escreva um programa que calcule o custo de uma viagem de carro.

O programa deve ler do usuário, usando `input`:

- a distância da viagem, em km (`float`);
- o consumo médio do carro, em km/litro (`float`);
- o preço do litro do combustível, em reais (`float`);
- o valor total dos pedágios, em reais (`float`);
- o número de pessoas que dividirão a despesa (`int`).

Em seguida o programa deve calcular e exibir:

1. a quantidade de litros necessária (distância dividida pelo consumo);
2. o custo do combustível;
3. o custo total (combustível + pedágios);
4. o custo por pessoa;
5. o tipo (`type`) das variáveis da distância e do número de pessoas.

Todos os valores em reais devem ser exibidos com **2 casas decimais** (use `round` ou formatação de string).

Exemplo:

```
Distância da viagem (km): 450
Consumo médio (km/l): 12.5
Preço do litro (R$): 5.89
Total de pedágios (R$): 32.40
Número de pessoas: 3

Litros necessários: 36.0
Custo do combustível: R$ 212.04
Custo total da viagem: R$ 244.44
Custo por pessoa: R$ 81.48
Tipo da distância: <class 'float'> | Tipo do número de pessoas: <class 'int'>
```

````python
"""Aguarda Entradas do usuario"""
distancia = float(input("Distância da viagem (km): "))
consumo = float(input("Consumo médio (km/l): "))
preco_litro = float(input("Preço do litro (R$): "))
pedagios = float(input("Total de pedágios (R$): "))
pessoas = int(input("Número de pessoas: "))

"""Realiza Calculos"""
litros = distancia / consumo
custo_combustivel = litros * preco_litro
custo_total = custo_combustivel + pedagios
custo_por_pessoa = custo_total / pessoas

"""Exibe Resultados"""
print(f"\nLitros necessários: {round(litros, 2)}")
print(f"Custo do combustível: R$ {custo_combustivel:.2f}")
print(f"Custo total da viagem: R$ {custo_total:.2f}")
print(f"Custo por pessoa: R$ {custo_por_pessoa:.2f}")
print(f"Tipo da distância: {type(distancia)} | Tipo do número de pessoas: {type(pessoas)}")
````

### Questão 3: Distância entre dois pontos (1,0 ponto)

Usando a instrução `import` e o módulo `math`, escreva um programa que:

1. leia as coordenadas de dois pontos do plano: `x1`, `y1`, `x2` e `y2` (todos `float`);
2. calcule `dx` e `dy` como as diferenças **em valor absoluto** entre as coordenadas;
3. calcule a distância entre os pontos com a fórmula abaixo:

$$distancia = \sqrt{dx^2 + dy^2}$$

4. calcule a área do círculo que tem essa distância como raio;
5. exiba `dx`, `dy`, a distância e a área, todos arredondados para 2 casas decimais com `round`.

Exemplo:

```
x1: 1
y1: 2
x2: 4
y2: 6

dx = 3.0
dy = 4.0
Distância entre os pontos: 5.0
Área do círculo de raio 5.0: 78.54
```

````python
import math

x1 = float(input("x1: "))
y1 = float(input("y1: "))
x2 = float(input("x2: "))
y2 = float(input("y2: "))

dx = abs(x2 - x1)
dy = abs(y2 - y1)
distancia = math.sqrt(dx ** 2 + dy ** 2)
area = math.pi * distancia ** 2

print("dx =", round(dx, 2))
print("dy =", round(dy, 2))
print("Distância entre os pontos:", round(distancia, 2))
print("Área do círculo de raio", round(distancia, 2), ":", round(area, 2))
````

### Questão 5: Generalização, parâmetros e variáveis locais (1,0 ponto)

**(a)** Escreva uma função chamada `moldura(texto, caractere, largura)` que exibe um texto dentro de uma moldura.
A função deve:

- imprimir uma linha formada pela repetição de `caractere`, com a `largura` especificada;
- imprimir o `texto` entre dois `caractere` (um no início e outro no fim da linha);
- imprimir novamente a linha de fechamento.

A função deve ter uma **docstring**. Note que ela **não retorna valor**, apenas exibe (retorna `None`).

Exemplo:

```
moldura("Trabalho 1", "*", 20)
```

```
********************
```

**(b)** Chame `moldura` **três vezes**, com textos, caracteres e larguras diferentes, mostrando que a função foi

````python
# @title
def moldura(texto, caractere, largura):
    """Exibe um texto entre duas linhas de caracteres."""
    linha = caractere * largura
    print(linha)
    print(caractere + " " + texto)
    print(linha)

moldura("Trabalho 1", "*", 20)
moldura("Python", "-", 15)
moldura("Olá, mundo!", "#", 18)
````

### Questão 7: Funções booleanas e operadores lógicos (1,0 ponto)

(a) Escreva uma função chamada eh_bissexto(ano) que retorne True se o ano for bissexto e False caso contrário. Use os operadores %, and, or e not em uma única expressão booleana no return (não use if).

(b) Escreva uma função chamada eh_triangulo(a, b, c) que retorne True se três medidas podem formar um triângulo. A condição é que cada lado seja positivo e que cada lado seja menor que a soma dos outros dois.

(c) Usando for i in range(...), exiba para os anos de 1996 a 2005 se cada um é bissexto ou não.

(d) Teste eh_triangulo com pelo menos três combinações diferentes, exibindo o resultado de cada teste.


````python
# Questão 7

# (a) Função para verificar se um ano é bissexto
def eh_bissexto(ano):
    return (ano % 400 == 0) or ((ano % 4 == 0) and not (ano % 100 == 0))


# (b) Função para verificar se três medidas formam um triângulo
def eh_triangulo(a, b, c):
    return (a > 0 and b > 0 and c > 0) and \
           (a < b + c and b < a + c and c < a + b)


# (c) Exibir os anos de 1996 a 2005
print("Anos bissextos de 1996 a 2005:")

for i in range(1996, 2006):
    print(i, eh_bissexto(i))


# (d) Testar três combinações de triângulos
print("\nTestes de triângulos:")

print(eh_triangulo(3, 4, 5))
print(eh_triangulo(1, 2, 3))
print(eh_triangulo(5, 5, 5))
````

### Questão 8: Condicionais encadeadas e aninhadas (1,0 ponto)

Escreva uma função chamada classificar_imc(peso, altura) que:

valide os dados: se peso ou altura for menor ou igual a zero, a função deve retornar None, sem calcular nada;
caso os dados sejam válidos, calcule o IMC com a fórmula imc = peso / altura ** 2;
use uma condicional encadeada (if/elif/else) para retornar a classificação correspondente:
| IMC | Classificação |
|---|---|
| Menor que 18,5 | Abaixo do peso |
| De 18,5 a 24,9 | Peso normal |
| De 25,0 a 29,9 | Sobrepeso |
| 30,0 ou mais | Obesidade |
Depois, escreva um programa que leia o peso e a altura do usuário, chame a função e:

se o retorno for None, exiba a mensagem "Dados inválidos: peso e altura devem ser positivos.";
caso contrário, exiba o IMC arredondado com 2 casas decimais e a classificação


````python
def classificar_imc(peso, altura):
    """Calcula o IMC e retorna sua classificação, ou None se inválido."""
    if peso <= 0 or altura <= 0:
        return None

    imc = peso / altura ** 2

    if imc < 18.5:
        return "Abaixo do peso"
    elif imc < 25:
        return "Peso normal"
    elif imc < 30:
        return "Sobrepeso"
    else:
        return "Obesidade"

peso = float(input("Peso (kg): "))
altura = float(input("Altura (m): "))
resultado = classificar_imc(peso, altura)

if resultado is None:
    print("Dados inválidos: peso e altura devem ser positivos.")
else:
    imc = peso / altura ** 2
    print(f"IMC: {imc:.2f} -> {resultado}")
````

### Questão 10: Boletim da turma (1,5 ponto)

Escreva um programa que gere o boletim de uma turma.

O programa deve:

1. ler a quantidade de alunos (`int`);
2. para cada aluno:
   - ler o nome (`str`) e três notas (`float`);
   - calcular a média com uma função `media_notas(nota1, nota2, nota3)` que **retorna** o valor;
   - obter a situação com uma função `situacao(media)` que **retorna** `"Aprovado"` (média maior ou igual a 7),
     `"Recuperação"` (média maior ou igual a 4 e menor que 7) ou `"Reprovado"` (média menor que 4);
   - exibir uma linha do boletim com o nome, a média (1 casa decimal) e a situação;
3. ao final, exibir:
   - a média geral da turma (2 casas decimais);
   - a quantidade e o **percentual** de aprovados;
   - o nome do aluno com a maior média.

Todas as funções devem ter **docstring**.

Exemplo:

```

Aluno 1 - nome: Ana

Nota 3: 9

Carla: 9.2 - Aprovado
```

**(a) Plano de desenvolvimento:** *antes* de programar, descreva na célula de
texto abaixo o seu plano: quais funções você vai criar, quais parâmetros cada uma recebe, o que cada uma retorna e em que ordem você vai construir o programa.

**(b) Desenvolvimento incremental:** escreva o programa em etapas, testando cada uma. Nas células de código indicadas, mantenha as **três versões**:

- **Versão 1:** apenas as funções `media_notas` e `situacao`, testadas com valores fixos (sem `input`).
- **Versão 2:** o laço que lê os alunos e exibe as linhas do boletim.
- **Versão 3:** a versão final, com as estatísticas do final do boletim.

**(c)** Ao final, comente na célula de texto o que mudou de uma versão para a outra e qual erro você encontrou (e corrigiu) durante o processo.

Primeiro vou criar e testar `media_notas`, depois `situacao`. Em seguida vou ler os alunos com um laço e guardar nome e média. Por fim, vou mostrar o boletim e calcular média geral, aprovados e melhor aluno.


````python
def media_notas(nota1, nota2, nota3):
    """Retorna a média de três notas."""
    return (nota1 + nota2 + nota3) / 3

def situacao(media):
    """Retorna a situação do aluno de acordo com a média."""
    if media >= 7:
        return "Aprovado"
    elif media >= 4:
        return "Recuperação"
    else:
        return "Reprovado"

print(media_notas(7, 5.5, 9))
print(situacao(7.17))
````

````python
quantidade = int(input("Quantos alunos? "))
alunos = []

for i in range(quantidade):
    nome = input(f"Aluno {i + 1} - nome: ")
    nota1 = float(input("Nota 1: "))
    nota2 = float(input("Nota 2: "))
    nota3 = float(input("Nota 3: "))
    media = media_notas(nota1, nota2, nota3)
    alunos.append((nome, media))

print("\n===== BOLETIM =====")
for nome, media in alunos:
    print(f"{nome}: {media:.1f} - {situacao(media)}")

medias = [media for nome, media in alunos]
media_geral = sum(medias) / quantidade
aprovados = sum(1 for media in medias if media >= 7)
melhor_nome, melhor_media = max(alunos, key=lambda aluno: aluno[1])

print("===================")
print(f"Média geral da turma: {media_geral:.2f}")
print(f"Aprovados: {aprovados} de {quantidade} ({aprovados / quantidade * 100:.2f}%)")
print(f"Melhor média: {melhor_nome} ({melhor_media:.2f})")
````
